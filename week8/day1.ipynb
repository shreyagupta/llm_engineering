# Note:
* In **`pricer_service`**, the bad version, every time the app goes to sleep (after just a minute or two), it must wake up, rebuild the image, redo imports, and reload the model—causing long delays on each call.
* **`pricer_service2`** improves this by using a **class-based structure** that allows pre-initialization steps to be done once instead of every time.
* It uses the **`@modal.build`** decorator to preload and cache the Hugging Face model files the first time the container is built.
* It uses **`@modal.enter`** to set up reusable components like the tokenizer and model once per container, so they persist across multiple invocations until the container sleeps again. So once this component containing the code to load the model, loads the model, it only does this once and it uses that for all the instances and calls. This is way more efficient!
* The **`price` method** then runs efficiently since all heavy setup work has already been done during build and enter phases.

### Notes for pricer_service2

### 🧩 What `@app.cls(...)` means

This is a **decorator** that tells Modal:

> “Hey, this class (`Pricer`) is a *remote class* that will run inside the Modal cloud environment, not locally.”

So, when you write:

```python
@app.cls(
    image=image.env({"HF_HUB_CACHE": CACHE_DIR}),
    secrets=secrets,
    gpu=GPU,
    timeout=1800,
    min_containers=MIN_CONTAINERS,
    volumes={CACHE_DIR: hf_cache_volume}
)
class Pricer:
    ...
```

You are **registering the class `Pricer` with your Modal app (`app`)**, and telling Modal *how* to run it in the cloud — what Docker image to use, what hardware, secrets, volumes, etc.

---

### 🏗️ Under the hood

When Modal sees `@app.cls(...)` on top of a class:

1. It **wraps** that class in a special Modal proxy.
2. It uses the arguments you passed (`image=...`, `gpu=...`, etc.) to define the **execution environment** (e.g., what container, GPU type, and secrets to use).
3. It registers that class definition with the Modal backend so that when you call it later, Modal knows how to spin up a container for it.

---

### ⚙️ Example: how it’s used

Later, you might do:

```python
pricer = Pricer()
result = pricer.price.remote("A pair of Nike shoes")
```

When you call `.remote()` on the method, Modal automatically:

* Spins up a container based on your `image` and environment configuration,
* Loads your class (`Pricer`) in that container,
* Calls the `setup()` method once (thanks to `@modal.enter()`),
* Executes the `price()` method with your argument,
* Returns the result back to you.

---

### 🧠 Analogy

Think of it like this:

| Local Python concept                | Modal equivalent                                              |
| ----------------------------------- | ------------------------------------------------------------- |
| `@dataclass`, `@staticmethod`, etc. | Local decorators that change Python behavior                  |
| `@app.cls(...)`                     | Registers a class as a **remote service** in the Modal cloud  |
| `@modal.method()`                   | Marks a method that can be **remotely invoked**               |
| `@modal.enter()`                    | Marks setup code that runs **once** when the container starts |

---

### 🪄 Summary

* `@app.cls` → turns your class into a **remote Modal class** (like a cloud microservice).
* The parameters (`image`, `gpu`, `volumes`, etc.) define its **environment**.
* `@modal.method` → marks a function that can be called **remotely**.
* `@modal.enter` → setup logic that runs **on container startup**.
